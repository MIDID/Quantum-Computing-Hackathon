## 1. Initializing the Game
The board starts with 9 empty cells.

```python
game = QuantumTicTacToe()
print("Current player:", game.current_player)
```

## 2. Placing Superposition Moves
Player X places $X_1$ across squares 0 and 1: $|\psi_1\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$.
Player O places $O_2$ across squares 1 and 2: $|\psi_2\rangle = \frac{1}{\sqrt{2}}(|1\rangle + |2\rangle)$.

```python
game.place_quantum_move(0, 1)  # X1
game.place_quantum_move(1, 2)  # O2
print("Uncollapsed moves:", game.uncollapsed_moves)
```

## 3. Creating Cyclic Entanglement
Player X places $X_3$ across squares 2 and 0. This closes an entanglement loop (0 - 1 - 2 - 0).

```python
success, msg, cycle_found = game.place_quantum_move(2, 0)  # X3
print("Cycle found:", cycle_found)
print(msg)
```

## 4. Quantum Circuit Execution & Wavefunction Collapse
Because an entanglement loop exists, the system collapses via Qiskit Aer simulation.

```python
collapse_msg, circuit = game.trigger_quantum_collapse()
print(collapse_msg)
print(circuit.draw(output="text"))
```
